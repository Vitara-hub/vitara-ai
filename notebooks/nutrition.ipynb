#Load Data

```python
import pandas as pd
```

##Nutrition Dataset

```python
calorie_df = pd.read_csv('https://drive.google.com/uc?id=1yy_U1Gwxjj6A9Iyu0eSLVZA2oe4TpUbz')
calorie_df['name'] = calorie_df['name'].str.lower()

calorie_df
```

##Classes

Download Class: https://drive.google.com/file/d/1oRg9D0svbfvz5nXzCZYFYiiRZkjf3Jhw/view?usp=sharing

```python
with open('classes.txt', 'r') as f:
    classes = [line.strip() for line in f]
```

#Mapping from Dataset

```python
mapping_keyword = {
    'Mie Goreng': 'mie goreng',
    'Nasi Goreng': 'nasi goreng',
    'Gudeg': 'gudeg',
    'Bakso': 'bakso',
    'Burger': 'burger',
    'Ayam Goreng': 'ayam goreng',
    'Capcay': 'cap cai sayur',
    'Rawon': 'rawon',
    'Rendang': 'rendang',
    'Nasi Uduk': 'nasi uduk',
    'French Fries': 'french fries',
    'Gado-Gado': 'gado gado',
    'Donat': 'donat',
    'Ayam Geprek': 'ayam geprek',
    'Pecel': 'pecel',
    'Bebek Betutu': 'bebek betutu',
    'Nasi Padang': 'nasi padang',
    'Pancake': 'pancake',
    'Soto': 'soto ayam',
    'Pizza': 'pizza',
    'Ikan Goreng': 'ikan mas goreng',
    'Pempek': 'pempek',
    'Sate': 'sate',
    'Tempe Goreng': 'tempe'
}
```

```python
result = []

for cls in classes:
    keyword = mapping_keyword.get(cls, cls.lower())

    subset = calorie_df[calorie_df['name'].str.contains(keyword, na=False)]

    if len(subset) > 0:
        calories = int(subset.iloc[0]['calories'])
    else:
        print(f"NOT FOUND: {cls}")
        calories = 0

    result.append({
        'class_name': cls,
        'calories_per_100g': calories
    })
```

#Manual Mapping

Data diambil dari situs fatsecret:
https://www.fatsecret.co.id/kalori-gizi/

```python
manual_fill = {
    'Donat': 412,
    'Ayam Geprek': 244,
    'Pecel': 244,
    'Bebek Betutu': 212,
    'Nasi Padang': 180,
    'Pancake': 520,
    'Pizza': 265,
    'French Fries': 156,
    'Gado-Gado': 132,
    'Soto': 130
}
```

```python
calorie_df = pd.DataFrame(result)
for i, row in calorie_df.iterrows():
    if row['calories_per_100g'] == 0:
        cls = row['class_name']
        if cls in manual_fill:
            calorie_df.at[i, 'calories_per_100g'] = manual_fill[cls]

print(calorie_df)
```

```python
calorie_df.to_csv('calorie_map.csv', index=False)

print("DONE")
print(calorie_df)
```
